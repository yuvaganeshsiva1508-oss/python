# keys

```python
mydict = {
    "Name" : "yuvaganesh",
    "age"  :  21,
    "city" : "kachipuram",
     500   : 60.0 , 

}

print(mydict)
print()
print(list(mydict.keys()))
print()


ans = list(mydict.keys())
print(ans[1])

print()
for each in mydict.keys():
    print(each) 

```

# **valuse** # ()

```python
mydict = {
    "Name" : "yuvaganesh",
    "age"  :  21,
    "city" : "kachipuram",
     500   : 60.0 , 

}

print(mydict)

print(list(mydict.values()))
print()
for each in mydict.values():
    print(each)
```

# **iteam** #()

```python
mydict = {
    "Name" : "yuvaganesh",
    "age"  :  21,
    "city" : "kachipuram",
     500   : 60.0 , 

}

print(mydict)

print()
print(list(mydict.items()))


print()
for each in mydict.items():
    print(f"{each[0]} \t {each[1]}")
```
