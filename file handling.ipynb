## ** file handling** #

```python
myfile = open("C:/Users/YUVAGANESH.S/OneDrive/Desktop/yuva.txt")

print(myfile.tell())

data = myfile.read(100)
print(data)
print(myfile.tell())

print('--------')
data = myfile.read(200)
print(data)
print(myfile.tell())

myfile.seek(0)
print(myfile.tell())

print('--------')

data = myfile.read(500)
print(data)
print(myfile.tell())


myfile.close()
```
