## **while  **<condution>:**##

```python
a =100
while a >=0:
  print(a, end=" ")
  a-=10


print("/nLOOP ENDS")
```

```python

a =1

while a>0:
    if a == 7:
        print(a)
        break
    print(a, end=" ")
    a +=1
```

**##** nested while loop***

```python
a = 1
while a<=3:
    b=1
    while b<=5:
        print(b, end=" ")
        b+=1
    a+=1
    print()
```
